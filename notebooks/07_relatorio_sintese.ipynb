## Objetivo
Consolidar todos os resultados em relatorio executivo de analise inflacionaria.

## Composicao
1. Resumo executivo
2. Dados e indicadores principais
3. Graficos consolidados
4. Analises complementares
5. Conclusoes

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from datetime import datetime

# Configuracoes
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14, 7)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
```

## 1. Carregar dados

```python
caminho_dados = Path('../dados/brutos/indicadores_consolidados.csv')
df = pd.read_csv(caminho_dados)
df['data'] = pd.to_datetime(df['data'])
df = df.sort_values('data').reset_index(drop=True)

print(f'Dados carregados: {len(df)} observacoes')
print(f'Periodo: {df["data"].min().strftime("%b/%Y")} a {df["data"].max().strftime("%b/%Y")}')
```

# RELATORIO EXECUTIVO - ANALISE DE INFLACAO BRASILEIRA

```python
print('='*80)
print('RELATORIO DE INFLACAO E INDICADORES ECONOMICOS DO BRASIL')
print('='*80)
print()
print(f'Data do Relatorio: {datetime.now().strftime("%d de %B de %Y")}')
print(f'Periodo Analisado: {df["data"].min().strftime("%B/%Y")} a {df["data"].max().strftime("%B/%Y")}')
print(f'Total de Observacoes: {len(df)} meses')
print()
print('Instituicao: FGV IBRE (Instituto Brasileiro de Economia)')
print('Desenvolvedor: Isabel Cruz (bellDataSc)')
```

## 2. Estatisticas Principais

```python
print('
' + '='*80)
print('INDICADORES PRINCIPAIS - ESTATISTICAS')
print('='*80)

# Tabela de resumo
resumo = pd.DataFrame({
    'IPCA Mensal': df['ipca_mensal'].describe(),
    'Desemprego': df['taxa_desemprego'].describe(),
    'Confianca': df['indice_confianca_fgv'].describe(),
    'Producao': df['variacao_producao_industrial'].describe()
})

print(resumo.to_string())
print()
print('Unidades: IPCA e Desemprego em %, Confianca em indice, Producao em %')
```

## 3. Indicadores Atuais (Ultimo periodo)

```python
ultimo = df.iloc[-1]
penultimo = df.iloc[-2] if len(df) > 1 else df.iloc[-1]

print('
' + '='*80)
print(f'INDICADORES ATUAIS ({ultimo["data"].strftime("%B/%Y")})')
print('='*80)
print()
print(f'IPCA Mensal:
      Valor: {ultimo["ipca_mensal"]:.3f}%
      Variacao: {ultimo["ipca_mensal"] - penultimo["ipca_mensal"]:+.3f} pp
      Media 12 meses: {ultimo["ipca_acumulado_12m"]:.2f}%')
print()
print(f'Taxa de Desemprego:
      Valor: {ultimo["taxa_desemprego"]:.2f}%
      Variacao: {ultimo["taxa_desemprego"] - penultimo["taxa_desemprego"]:+.2f} pp')
print()
print(f'Indice de Confianca do Consumidor:
      Valor: {ultimo["indice_confianca_fgv"]:.1f}
      Variacao: {ultimo["indice_confianca_fgv"] - penultimo["indice_confianca_fgv"]:+.1f} pts')
print()
print(f'Producao Industrial:
      Variacao: {ultimo["variacao_producao_industrial"]:.2f}%')
```

## 4. Analise de Correlacoes

```python
# Matriz de correlacao
correlacoes = df[['ipca_mensal', 'taxa_desemprego', 'indice_confianca_fgv', 'variacao_producao_industrial']].corr()

print('
' + '='*80)
print('MATRIZ DE CORRELACOES')
print('='*80)
print(correlacoes.to_string())
print()
print('Interpretacao:')
print(f'  IPCA vs Desemprego: {correlacoes.loc["ipca_mensal", "taxa_desemprego"]:+.3f} (Correlacao {"fraca negativa" if correlacoes.loc["ipca_mensal", "taxa_desemprego"] < -0.3 else "moderada/fraca"})')
print(f'  IPCA vs Confianca: {correlacoes.loc["ipca_mensal", "indice_confianca_fgv"]:+.3f}')
print(f'  IPCA vs Producao: {correlacoes.loc["ipca_mensal", "variacao_producao_industrial"]:+.3f}')
```

## 5. Graficos Consolidados

```python
fig = plt.figure(figsize=(16, 12))

# 1. IPCA Mensal
ax1 = plt.subplot(2, 2, 1)
ax1.plot(df['data'], df['ipca_mensal'], linewidth=2, color='#d62728', marker='o', markersize=3)
ax1.fill_between(df['data'], df['ipca_mensal'], alpha=0.3, color='#d62728')
ax1.axhline(y=df['ipca_mensal'].mean(), color='black', linestyle='--', linewidth=1.5, alpha=0.7)
ax1.set_title('IPCA Mensal', fontsize=12, fontweight='bold')
ax1.set_ylabel('IPCA (%)')
ax1.grid(True, alpha=0.3)

# 2. Desemprego
ax2 = plt.subplot(2, 2, 2)
ax2.plot(df['data'], df['taxa_desemprego'], linewidth=2, color='#1f77b4', marker='s', markersize=3)
ax2.fill_between(df['data'], df['taxa_desemprego'], alpha=0.3, color='#1f77b4')
ax2.set_title('Taxa de Desemprego', fontsize=12, fontweight='bold')
ax2.set_ylabel('Taxa (%)')
ax2.grid(True, alpha=0.3)

# 3. Confianca
ax3 = plt.subplot(2, 2, 3)
ax3.plot(df['data'], df['indice_confianca_fgv'], linewidth=2, color='#2ca02c', marker='^', markersize=3)
ax3.fill_between(df['data'], df['indice_confianca_fgv'], alpha=0.3, color='#2ca02c')
ax3.set_title('Indice de Confianca (FGV)', fontsize=12, fontweight='bold')
ax3.set_ylabel('Indice')
ax3.set_xlabel('Data')
ax3.grid(True, alpha=0.3)

# 4. Producao Industrial
ax4 = plt.subplot(2, 2, 4)
ax4.plot(df['data'], df['variacao_producao_industrial'], linewidth=2, color='#ff7f0e', marker='D', markersize=3)
ax4.axhline(y=0, color='black', linewidth=1)
ax4.fill_between(df['data'], df['variacao_producao_industrial'], alpha=0.3, color='#ff7f0e')
ax4.set_title('Variacao da Producao Industrial', fontsize=12, fontweight='bold')
ax4.set_ylabel('Variacao (%)')
ax4.set_xlabel('Data')
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 6. Heatmap de Correlacoes

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlacoes, annot=True, fmt='.3f', cmap='coolwarm', center=0,
            cbar_kws={'label': 'Correlacao'}, vmin=-1, vmax=1, linewidths=1)
plt.title('Matriz de Correlacoes - Indicadores Economicos', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 7. Conclusoes e Insights

```python
print('
' + '='*80)
print('CONCLUSOES E INSIGHTS PRINCIPAIS')
print('='*80)
print()
print('1. TENDENCIA INFLACIONARIA:')
ipca_atual = df['ipca_mensal'].iloc[-1]
ipca_media = df['ipca_mensal'].mean()
tendencia = 'ACIMA' if ipca_atual > ipca_media else 'ABAIXO'
print(f'   Inflacao atual ({ipca_atual:.3f}%) esta {tendencia} da media historica ({ipca_media:.3f}%)')
print(f'   IPCA acumulado 12 meses: {df["ipca_acumulado_12m"].iloc[-1]:.2f}%')
print()
print('2. MERCADO DE TRABALHO:')
desemprego = df['taxa_desemprego'].iloc[-1]
desemprego_media = df['taxa_desemprego'].mean()
print(f'   Taxa de desemprego: {desemprego:.2f}%')
print(f'   Media historica: {desemprego_media:.2f}%')
print(f'   Diferenca: {desemprego - desemprego_media:+.2f} pp')
print()
print('3. CONFIANCA DO CONSUMIDOR:')
confianca = df['indice_confianca_fgv'].iloc[-1]
confianca_media = df['indice_confianca_fgv'].mean()
print(f'   Indice atual: {confianca:.1f}')
print(f'   Media historica: {confianca_media:.1f}')
print(f'   Avaliacao: {"Acima da media" if confianca > confianca_media else "Abaixo da media"}')
print()
print('4. ATIVIDADE ECONOMICA:')
producao = df['variacao_producao_industrial'].iloc[-1]
print(f'   Producao industrial: {producao:+.2f}%')
print(f'   Tendencia: {"Expansao" if producao > 0 else "Contracao"}')
print()
print('5. RECOMENDACOES:')
if ipca_atual > ipca_media + 0.1:
    print('   [!] Inflacao acima da media - Monitoramento recomendado')
if desemprego > desemprego_media + 1:
    print('   [!] Desemprego elevado - Politicas de emprego importante')
if confianca < confianca_media:
    print('   [!] Confianca abaixo da media - Recuperacao de credibilidade necessaria')
```
