# 01 - Carregamento e Exploracao de Dados

## Objetivo
Explorar os dados coletados, entender as dimensoes e fazer analise descritiva inicial.

## Fluxo
1. Carregar dados consolidados
2. Explorar dimensoes e tipos
3. Estatisticas descritivas
4. Visualizacoes iniciais

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Configurar visualizacoes
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Carregar dados consolidados

```python
# Caminho do arquivo consolidado
caminho_dados = Path('../dados/brutos/indicadores_consolidados.csv')

# Carregar dados
df = pd.read_csv(caminho_dados)

# Converter coluna data para datetime
df['data'] = pd.to_datetime(df['data'])

print(f'Dados carregados com sucesso!')
print(f'Dimensoes: {df.shape[0]} linhas x {df.shape[1]} colunas')
```

## 2. Explorar estrutura dos dados

```python
# Tipos de dados
print('TIPOS DE DADOS:')
print(df.dtypes)
print()

# Primeiras linhas
print('PRIMEIRAS 5 LINHAS:')
print(df.head())
print()

# Informacoes gerais
print('INFORMACOES GERAIS:')
df.info()
```

## 3. Periodo temporal

```python
# Periodo coberto
data_inicio = df['data'].min()
data_fim = df['data'].max()
diferenca_meses = (data_fim.year - data_inicio.year) * 12 + (data_fim.month - data_inicio.month)

print(f'Periodo: {data_inicio.date()} a {data_fim.date()}')
print(f'Duracao: {diferenca_meses + 1} meses')
```

## 4. Estatisticas descritivas

```python
# Resumo estatistico
print('RESUMO ESTATISTICO:')
print(df.describe())
```

## 5. Valores ausentes

```python
# Verificar valores ausentes
print('VALORES AUSENTES:')
ausentes = df.isnull().sum()
print(ausentes)

if ausentes.sum() == 0:
    print('
Nao ha valores ausentes!')
```

## 6. Visualizacoes das series temporais

```python
# Criar figura com 4 subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Series Temporais - Indicadores Economicos', fontsize=16, fontweight='bold')

# IPCA Mensal
axes[0, 0].plot(df['data'], df['ipca_mensal'], linewidth=2, color='#1f77b4')
axes[0, 0].set_title('IPCA Mensal (%)')
axes[0, 0].set_ylabel('Percentual (%)')
axes[0, 0].grid(True, alpha=0.3)

# IPCA 12 meses
axes[0, 1].plot(df['data'], df['ipca_acumulado_12_meses'], linewidth=2, color='#ff7f0e')
axes[0, 1].set_title('IPCA Acumulado 12 Meses (%)')
axes[0, 1].set_ylabel('Percentual (%)')
axes[0, 1].grid(True, alpha=0.3)

# Desemprego
axes[1, 0].plot(df['data'], df['taxa_desemprego'], linewidth=2, color='#2ca02c')
axes[1, 0].set_title('Taxa de Desemprego (%)')
axes[1, 0].set_ylabel('Percentual (%)')
axes[1, 0].grid(True, alpha=0.3)

# Producao Industrial
axes[1, 1].plot(df['data'], df['variacao_producao_industrial'], linewidth=2, color='#d62728')
axes[1, 1].set_title('Variacao Producao Industrial (%)')
axes[1, 1].set_ylabel('Percentual (%)')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 7. Correlacao entre variaveis

```python
# Matriz de correlacao
colunas_numericas = df.select_dtypes(include=[np.number]).columns
correlacao = df[colunas_numericas].corr()

# Visualizar correlacao
plt.figure(figsize=(8, 6))
sns.heatmap(correlacao, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Matriz de Correlacao - Indicadores Economicos')
plt.tight_layout()
plt.show()
```

## 8. Resumo final

```python
print('RESUMO DA EXPLORACAO:')
print('=' * 60)
print(f'Periodo: {data_inicio.strftime("%B %Y")} a {data_fim.strftime("%B %Y")}')
print(f'Observacoes: {len(df)} meses')
print(f'Variaveis: {len(colunas_numericas)}')
print()
print('MEDIAS:')
print(f'  IPCA Mensal: {df["ipca_mensal"].mean():.3f}%')
print(f'  Desemprego: {df["taxa_desemprego"].mean():.2f}%')
print(f'  Confianca: {df["indice_confianca_consumidor"].mean():.2f}')
print(f'  Producao Industrial: {df["variacao_producao_industrial"].mean():.2f}%')
print()
print('Dados prontos para proxima etapa (validacao de qualidade)')
```
